ejercicio 6.1

```python
productos = []
ejecutando = True

print("--- Registro de Inventario (Escriba 'fin' para terminar) ---")

while ejecutando:
    nombre_producto = input("\nIngrese nombre del producto: ").strip()

    if nombre_producto.lower() == 'fin':
        ejecutando = False
        break

    precio_valido = False
    precio = 0.0

    while not precio_valido:
        entrada_precio = input(f"Ingrese precio para '{nombre_producto}': ")

        try:
            precio = float(entrada_precio)
            if precio < 0:
                print(" Error: El precio no puede ser negativo.")
                continue
            precio_valido = True
        except ValueError:
            print(f" Error: '{entrada_precio}' no es un número válido.")

    cantidad_valida = False
    cantidad = 0

    while not cantidad_valida:
        entrada_cantidad = input(f"Ingrese cantidad para '{nombre_producto}': ")

        try:
            cantidad = int(entrada_cantidad)
            if cantidad < 0:
                print(" Error: La cantidad no puede ser negativa.")
                continue
            cantidad_valida = True
        except ValueError:
            print(f" Error: '{entrada_cantidad}' no es un número entero.")

    productos.append({
        "nombre": nombre_producto,
        "precio": precio,
        "cantidad": cantidad
    })
    print(f"Producto '{nombre_producto}' registrado exitosamente.")


print("\n--- Estado Final del Inventario Registrado ---")
if productos:
    total_inventario = 0.0
    for prod in productos:
        subtotal = prod["precio"] * prod["cantidad"]
        total_inventario += subtotal
        print(f" - {prod['nombre']}: {prod['cantidad']} unds x ${prod['precio']:.2f} = ${subtotal:.2f}")
    print(f"\nValor total acumulado: ${total_inventario:.2f}")
else:
    print("No se registraron productos.")
```

```python
import json


def guardar_inventario_csv(inventario, ruta_archivo="inventario.csv"):
    archivo = open(ruta_archivo, "w", encoding="utf-8")

    archivo.write("nombre,precio,stock\n")

    for producto in inventario:
        linea = f"{producto['nombre']},{producto['precio']},{producto['stock']}\n"
        archivo.write(linea)

    archivo.close()
    print("Inventario guardado correctamente en CSV.")


def cargar_inventario_csv(ruta_archivo="inventario.csv"):
    try:
        archivo = open(ruta_archivo, "r", encoding="utf-8")
        lineas = archivo.readlines()
        archivo.close()

        inventario = []

        for i in range(1, len(lineas)):
            linea_limpia = lineas[i].strip()
            if linea_limpia != "":
                partes = linea_limpia.split(",")
                producto = {
                    "nombre": partes[0],
                    "precio": float(partes[1]),
                    "stock": int(partes[2])
                }
                inventario.append(producto)

        return inventario

    except FileNotFoundError:
        print("El archivo CSV no existe. Iniciando con inventario vacío.")
        return []




def guardar_inventario_json(inventario, ruta_archivo="inventario.json"):
    archivo = open(ruta_archivo, "w", encoding="utf-8")
    json.dump(inventario, archivo, indent=4)
    archivo.close()
    print("Inventario guardado correctamente en JSON.")


def cargar_inventario_json(ruta_archivo="inventario.json"):
    try:
        archivo = open(ruta_archivo, "r", encoding="utf-8")
        inventario = json.load(archivo)
        archivo.close()
        return inventario
    except FileNotFoundError:
        print("El archivo JSON no existe. Iniciando con inventario vacío.")
        return []

inventario_actual = cargar_inventario_csv("inventario_no_existente.csv")


inventario_actual = [
    {"nombre": "Pan", "precio": 1500.0, "stock": 30},
    {"nombre": "Leche", "precio": 3500.0, "stock": 13},
    {"nombre": "Café", "precio": 8000.0, "stock": 10}
]

guardar_inventario_csv(inventario_actual)
inventario_desde_csv = cargar_inventario_csv("inventario.csv")
print("Inventario reconstruido desde CSV:", inventario_desde_csv)

guardar_inventario_json(inventario_actual)
inventario_desde_json = cargar_inventario_json("inventario.json")
print("Inventario reconstruido desde JSON:", inventario_desde_json)
```

### Ventajas del formato JSON frente a CSV

1. **Preservación de tipos de datos nativos:**
   * **CSV:** Almacena todo el contenido como texto plano. Al leer el archivo, es obligatorio realizar conversiones manuales mediante código (`float()`, `int()`) para restaurar los tipos numéricos.
   * **JSON:** Mantiene de forma nativa la diferencia entre cadenas de texto (`string`), enteros (`int`), flotantes (`float`), booleanos (`bool`) y valores nulos (`None`).

2. **Soporte para estructuras jerárquicas y anidadas:**
   * **CSV:** Está limitado a una estructura tabular plana de dos dimensiones (filas y columnas). Resulta complejo o imposible representar objetos anidados de manera limpia (por ejemplo, si un producto tuviera una lista de ingredientes o un diccionario con especificaciones técnicas).
   * **JSON:** Permite anidar listas y diccionarios con múltiples niveles de profundidad sin perder la relación entre los datos.

3. **Inmunidad a problemas con delimitadores:**
   * **CSV:** Si el valor de un campo de texto contiene la comilla o la coma empleada como separador (por ejemplo, un nombre como `"Pan, integral"`), el archivo puede corromperse al momento de hacer el `.split(",")`.
   * **JSON:** Gestiona automáticamente el escape de caracteres especiales y comillas, garantizando la integridad de las cadenas.

4. **Facilidad de serialización/deserialización:**
   * Con las funciones de la librería estándar `json.dump()` y `json.load()`, no se requiere escribir bucles manuales de parsing ni manejar caracteres de salto de línea (`\n`), reduciendo el margen de error en el mantenimiento del código.
