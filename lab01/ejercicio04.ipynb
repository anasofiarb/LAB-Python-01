```python

contador_global = 0

def incrementar_global():
    global contador_global
    contador_global = contador_global + 1

incrementar_global()
print("Contador (usando global):", contador_global)

contador_global = 0

def incrementar_puro(valor_actual):
    return valor_actual + 1

contador_global = incrementar_puro(contador_global)
```

* **Error obtenido:**
  `UnboundLocalError: local variable 'contador_global' referenced before assignment`

* **¿Por qué ocurre?**
  Cuando Python analiza el cuerpo de una función y detecta una operación de asignación (`contador_global = ...`), asume por defecto que la variable pertenece al **ámbito local** de esa función. Al intentar evaluar el lado derecho de la asignación (`contador_global + 1`), intenta leer el valor de la variable local `contador_global`, pero esta aún no tiene ningún valor asignado en ese entorno local, generando el error.

---

### Comparación de Proprobabilidad (Testabilidad)

La **Solución 2 (pasar el valor por parámetro y retornar el resultado)** es significativamente más fácil de probar y mantener por las siguientes razones:

1. **Aislamiento y modularidad:** La función no depende del estado externo ni de variables globales. Recibe un dato de entrada, procesa una salida y no altera el entorno global (ausencia de efectos secundarios).
2. **Facilidad en pruebas unitarias:** Se puede invocar directamente en cualquier entorno de pruebas enviando cualquier número de entrada (`incrementar_puro(10)`), verificando de inmediato que retorna `11`, sin necesidad de inicializar, modificar o limpiar variables globales antes o después de cada prueba.
3. **Control de efectos secundarios:** Modificar variables globales directamente con `global` (Solución 1) puede provocar errores encadenados difíciles de rastrear cuando múltiples funciones intentan leer o sobreescribir la misma memoria.

```python
import datetime


cuenta = {
    "titular": "Ana",
    "saldo": 200000.0,
    "movimientos": []
}


def consultar_saldo(c):
    print(f"\n--- Consulta de Saldo ---")
    print(f"Titular: {c['titular']}")
    print(f"Saldo actual: ${c['saldo']:.2f}")


def depositar(c, monto):
    if monto > 0:
        c["saldo"] += monto
        fecha_actual = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        movimiento = {
            "tipo": "Depósito",
            "monto": monto,
            "saldo_resultante": c["saldo"],
            "fecha": fecha_actual
        }
        c["movimientos"].append(movimiento)
        print(f"Depósito exitoso de ${monto:.2f}.")
    else:
        print("Error: El monto a depositar debe ser positivo.")


def retirar(c, monto):
    if monto <= 0:
        print("Error: El monto a retirar debe ser positivo.")
    elif monto > c["saldo"]:
        print(f"Error: Saldo insuficiente (${c['saldo']:.2f}) para retirar ${monto:.2f}.")
    else:
        c["saldo"] -= monto
        fecha_actual = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        movimiento = {
            "tipo": "Retiro",
            "monto": monto,
            "saldo_resultante": c["saldo"],
            "fecha": fecha_actual
        }
        c["movimientos"].append(movimiento)
        print(f"Retiro exitoso de ${monto:.2f}.")


def mostrar_historial(c):
    print(f"\n--- Historial de Movimientos de {c['titular']} ---")
    if not c["movimientos"]:
        print("No hay movimientos registrados.")
    else:
        for mov in c["movimientos"]:
            print(f"[{mov['fecha']}] {mov['tipo']}: ${mov['monto']:.2f} | Saldo: ${mov['saldo_resultante']:.2f}")



consultar_saldo(cuenta)
depositar(cuenta, 50000)
retirar(cuenta, 30000)
retirar(cuenta, 300000)
mostrar_historial(cuenta)
consultar_saldo(cuenta)
```
