```python
import requests


url = "https://jsonplaceholder.typicode.com/users"
response = requests.get(url)
usuarios = response.json()

nombres_filtrados = []


for usuario in usuarios:
    correo = usuario["email"]
    if correo.endswith(".biz"):
        nombres_filtrados.append(usuario["name"])


print("Usuarios cuyo correo termina en .biz:")
for nombre in nombres_filtrados:
    print("-", nombre)
```

```python
import requests

nombre_pokemon_entrada = input("Ingrese el nombre de un Pokémon: ")
nombre_pokemon = nombre_pokemon_entrada.lower().strip()

url = f"https://pokeapi.co/api/v2/pokemon/{nombre_pokemon}"

try:
    respuesta = requests.get(url)
except requests.exceptions.RequestException:
    print("Error de conexión al intentar consultar la base de datos de Pokémon.")
    exit()

if respuesta.status_code == 200:
    datos = respuesta.json()

    lista_tipos = []
    lista_datos_tipos = datos["types"]
    for diccionario_tipo in lista_datos_tipos:
        nombre_tipo = diccionario_tipo["type"]["name"]
        lista_tipos.append(nombre_tipo)

    lista_habilidades = []
    lista_datos_habilidades = datos["abilities"]
    for diccionario_habilidad in lista_datos_habilidades:
        nombre_habilidad = diccionario_habilidad["ability"]["name"]
        lista_habilidades.append(nombre_habilidad)

    pokemon = {
        "nombre": datos["name"],
        "altura": datos["height"],
        "peso": datos["weight"],
        "tipos": lista_tipos,
        "habilidades": lista_habilidades
    }


    print("\n Información del Pokémon")
    print(f"Nombre: {pokemon['nombre'].capitalize()}")
    print(f"Altura: {pokemon['altura']}")
    print(f"Peso: {pokemon['peso']}")

    tipos_formateados = ', '.join(pokemon['tipos'])
    print(f"Tipos: {tipos_formateados}")

    print("\n Clasificación:")


    cantidad_tipos = len(pokemon['tipos'])
    if cantidad_tipos > 1:
        print(" - Pokémon de tipo mixto.")
    elif cantidad_tipos == 1:
        print(" - Pokémon de tipo puro.")
    else:
        print(" - Sin tipo registrado.")

    peso_pokemon = pokemon['peso']
    if peso_pokemon > 100:
        print(" - Pokémon de gran tamaño.")
    else:
        print(" - Pokémon ligero.")

    print("\n Habilidades:")
    if not pokemon['habilidades']:
        print(" - Ninguna habilidad registrada.")
    else:
        for habilidad in pokemon['habilidades']:

            print(f" - {habilidad.replace('-', ' ').title()}")


    print("\n Algunos Movimientos (Primeros 5):")

    lista_datos_movimientos = datos["moves"]

    if not lista_datos_movimientos:
        print(" - No tiene movimientos registrados.")
    else:

        contador_movimientos = 0

        for diccionario_movimiento in lista_datos_movimientos:
            if contador_movimientos < 5:
                nombre_movimiento = diccionario_movimiento["move"]["name"]

                print(f" - {nombre_movimiento.replace('-', ' ').capitalize()}")
                contador_movimientos += 1
            else:
                break


else:
    if respuesta.status_code == 404:
        print(f"Pokémon '{nombre_pokemon_entrada}' no encontrado. Verifique el nombre ingresado.")
    else:
        print(f"Error al consultar la API. Código de estado: {respuesta.status_code}")
```
