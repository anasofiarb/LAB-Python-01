```python

n = 10
suma_naturales = 0

print(f"Evaluando números del 1 al {n}:")
for i in range(1, n + 1):

    suma_naturales += i


    if i % 2 == 0:
        print(f"Número par encontrado: {i}")

print(f"\nSuma total de los primeros {n} números naturales: {suma_naturales}")
```

```python

temperaturas = [22.5, 23.0, 21.8, 24.3, 25.1]


suma_temperaturas = 0.0
for temp in temperaturas:
    suma_temperaturas += temp

promedio = suma_temperaturas / len(temperaturas)


temperaturas_mayores = []
for temp in temperaturas:
    if temp > promedio:
        temperaturas_mayores.append(temp)


print(f"Promedio de temperaturas: {promedio:.2f}°C")
print("Temperaturas mayores al promedio:")
for temp in temperaturas_mayores:
    print(f" - {temp}°C")
```

```python

suma_pares = 0
suma_impares = 0

for i in range(1, 11):
    if i % 2 == 0:
        suma_pares += i
    else:
        suma_impares += i

estado_final = {
    "suma_pares": suma_pares,
    "suma_impares": suma_impares
}

print("Estado final del sistema:", estado_final)
```

### Explicación del cambio de estado en la memoria

* **Estado inicial (antes del bucle):**
  Las variables acumuladoras se crean e inicializan en memoria con valor de `0`: `suma_pares = 0` y `suma_impares = 0`.

* **Evolución iteración por iteración:**
  En cada paso del ciclo `for`, la variable de control `i` toma un entero secuencial del 1 al 10. La estructura condicional evalúa el residuo de la división entre 2 (`i % 2 == 0`):
  1. **Evaluación condicional:** Determina qué rama del flujo secuencial se ejecuta.
  2. **Reasignación de memoria:**
     * Si `i` es par, se evalúa `suma_pares + i` y el resultado reemplaza el valor anterior de `suma_pares`.
     * Si `i` es impar (bloque `else`), se evalúa `suma_impares + i` y el resultado reemplaza el valor anterior de `suma_impares`.

---

#### Tabla de trazabilidad de memoria

| Iteración (`i`) | Condición (`i % 2 == 0`) | Acción realizada | `suma_pares` | `suma_impares` |
| :--- | :--- | :--- | :--- | :--- |
| **Inicio** | — | Inicialización de variables | `0` | `0` |
| **`i = 1`** | `False` | `suma_impares += 1` | `0` | `1` |
| **`i = 2`** | `True` | `suma_pares += 2` | `2` | `1` |
| **`i = 3`** | `False` | `suma_impares += 3` | `2` | `4` |
| **`i = 4`** | `True` | `suma_pares += 4` | `6` | `4` |
| **`i = 5`** | `False` | `suma_impares += 5` | `6` | `9` |
| **`i = 6`** | `True` | `suma_pares += 6` | `12` | `9` |
| **`i = 7`** | `False` | `suma_impares += 7` | `12` | `16` |
| **`i = 8`** | `True` | `suma_pares += 8` | `20` | `16` |
| **`i = 9`** | `False` | `suma_impares += 9` | `20` | `25` |
| **`i = 10`** | `True` | `suma_pares += 10` | `30` | `25` |

---

* **Consolidación del estado final:**
  Al finalizar el ciclo, los valores resultantes en memoria (`30` y `25`) se extraen para construir el diccionario `estado_final`, registrando la fotografía definitiva de la ejecución.

```python
productos = []
print("Registro de inventario (escriba 'fin' para terminar).")

while True:
    nombre = input("Ingrese producto: ").strip()
    if nombre.lower() == "fin":
        break


    try:
        entrada_precio = input(f"Precio para '{nombre}': ")
        precio = float(entrada_precio)

        entrada_cantidad = input(f"Cantidad para '{nombre}': ")
        cantidad = int(entrada_cantidad)
    except ValueError:
        print("Error: Ingrese valores numéricos válidos. Reintentando producto.")
        continue
    if precio < 0 or cantidad < 0:
        print(f"Error: No se permiten precios ({precio}) o cantidades ({cantidad}) negativas. Registro saltado.")
        continue
    productos.append({"nombre": nombre, "precio": precio, "cantidad": cantidad})


-
total = 0.0
mas_caro = None

for p in productos:

    total += p["precio"] * p["cantidad"]

    if mas_caro is None:
        mas_caro = p

    elif p["precio"] > mas_caro["precio"]:
        mas_caro = p


print("\n--- Resultados del Inventario ---")

if productos:
    print(f"Total del inventario: ${total:.2f}")

    if mas_caro:
        def obtener_precio(p):
            return p["precio"]

        producto_mas_caro_max = max(productos, key=obtener_precio)

        print(f"Producto más caro (usando max): {producto_mas_caro_max['nombre']} (${producto_mas_caro_max['precio']:.2f})")


    print("\nEstado final del inventario (Lista de Diccionarios):")

    for item in productos:
        print(f" - {item}")
else:
    print("El inventario está vacío.")
```
