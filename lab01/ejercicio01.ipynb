```python

saldo = float(input("Ingrese el saldo inicial de la cuenta: "))
deposito = float(input("Ingrese el monto a depositar: "))
retiro = float(input("Ingrese el monto a retirar: "))

interes = 0.02

saldo = saldo + deposito
saldo = saldo - retiro

saldo_final = saldo + (saldo * interes)

print(f"Saldo final con interés: ${saldo_final:.2f}")
```

##Representación del Estado

Estado inicial: Registro en la RAM inmediatamente después de capturar los datos mediante input(), conteniendo los valores financieros de partida antes de cualquier operación.   


Estado final: Fotografía consolidada en memoria tras ejecutar los cálculos (saldo_final), reflejando el capital definitivo con el depósito, el retiro y los intereses aplicados.   



### Impacto de cada instrucción en la memoria RAM

1. **`saldo = float(input(...))` / `deposito = float(input(...))` / `retiro = float(input(...))`**
   * El intérprete pausa la ejecución esperando la entrada por teclado.
   * `float()` convierte la cadena de texto ingresada a un número de punto flotante de 64 bits.
   * Se reserva un espacio en memoria RAM para almacenar el valor y se asigna el identificador de variable correspondiente.

2. **`interes = 0.02`**
   * Asigna en el marco de variables la dirección de memoria donde se aloja el flotante `0.02`.

3. **`saldo = saldo + deposito`**
   * Lee los valores de `saldo` y `deposito` desde la memoria.
   * La ALU procesa la suma y crea un nuevo objeto flotante en memoria.
   * La variable `saldo` se reasigna a la nueva dirección (el valor anterior queda listo para el recolector de basura).

4. **`saldo = saldo - retiro`**
   * Calcula la resta con el valor actualizado de `saldo` y el valor de `retiro`.
   * Genera un nuevo resultado en memoria y reasigna el identificador `saldo`.

5. **`saldo_final = saldo + (saldo * interes)`**
   * Evalúa la multiplicación `saldo * interes` en la CPU.
   * Suma el resultado al `saldo` actual y crea una nueva posición en memoria identificada como `saldo_final`.

6. **`print(f"Saldo final con interés: ${saldo_final:.2f}")`**
   * Lee la variable `saldo_final` sin modificar el estado de la memoria.
   * Formatea el valor a 2 decimales y lo transmite a la salida estándar (`stdout`).

```python

temperaturas = [22.5, 23.0, 21.8, 24.3, 25.1]
print("Temperaturas iniciales:", temperaturas)

temperaturas.append(26.0)
print("Tras agregar 26.0 al final:", temperaturas)

temperaturas.pop(0)
print("Tras eliminar la primera posición:", temperaturas)

temperaturas.insert(1, 23.7)
print("Tras insertar 23.7 en la 2da posición:", temperaturas)

temperaturas.sort()
print("Lista ordenada de menor a mayor:", temperaturas)

temperatura_minima = temperaturas[0]
temperatura_maxima = temperaturas[0]
suma_total = 0.0

for temp in temperaturas:
    if temp < temperatura_minima:
        temperatura_minima = temp
    if temp > temperatura_maxima:
        temperatura_maxima = temp
    suma_total += temp

diferencia_temperaturas = temperatura_maxima - temperatura_minima
promedio_temperaturas = suma_total / len(temperaturas)

print(f"Diferencia entre máxima ({temperatura_maxima}) y mínima ({temperatura_minima}): {diferencia_temperaturas:.2f}°C")
print(f"Promedio de temperaturas: {promedio_temperaturas:.2f}°C")


posicion_23_7 = -1
for i in range(len(temperaturas)):
    if temperaturas[i] == 23.7:
        posicion_23_7 = i
        break

print(f"La temperatura 23.7 se encuentra en el índice: {posicion_23_7}")

sublista_2da_a_4ta = temperaturas[1:4]
print("Temperaturas de la 2da a la 4ta posición:", sublista_2da_a_4ta)

tres_primeras = temperaturas[:3]
print("Las tres primeras temperaturas:", tres_primeras)
```

ejercicio1.3

```python

inventario = [
    {"nombre": "Pan", "precio": 1500, "stock": 30},
    {"nombre": "Leche", "precio": 3500, "stock": 15},
    {"nombre": "Café", "precio": 8000, "stock": 10},
]


primer_producto = inventario[0]["nombre"]
print("Nombre del primer producto:", primer_producto)


nuevo_producto = {"nombre": "Huevos", "precio": 600, "stock": 0}
inventario.append(nuevo_producto)
print("Inventario tras agregar 'Huevos':", inventario)


for producto in inventario:
    if producto["nombre"] == "Leche":
        producto["stock"] -= 2
        break

valor_total_inventario = 0.0
for producto in inventario:
    valor_total_inventario += producto["precio"] * producto["stock"]

print(f"Valor total del inventario: ${valor_total_inventario:.2f}")


print("Productos con stock menor a 10:")
for producto in inventario:
    if producto["stock"] < 10:
        print(f" - {producto['nombre']}: stock = {producto['stock']}")


for producto in inventario:
    precio_aumentado = producto["precio"] * 1.10
    producto["precio"] = round(precio_aumentado, 2)

print("Inventario con precios actualizados (+10%):", inventario)


nombre_buscado = input("Ingrese el nombre del producto a buscar: ")
producto_encontrado = None

for producto in inventario:
    if producto["nombre"].lower() == nombre_buscado.lower():
        producto_encontrado = producto
        break

if producto_encontrado is not None:
    print("Producto encontrado:", producto_encontrado)
else:
    print(f"El producto '{nombre_buscado}' no se encuentra en el inventario.")


indice = 0
while indice < len(inventario):
    if inventario[indice]["stock"] == 0:
        producto_eliminado = inventario.pop(indice)
        print(f"Producto eliminado por stock 0: {producto_eliminado['nombre']}")
    else:
        indice += 1


n = len(inventario)
for i in range(n):
    for j in range(0, n - i - 1):
        if inventario[j]["precio"] > inventario[j + 1]["precio"]:

            temporal = inventario[j]
            inventario[j] = inventario[j + 1]
            inventario[j + 1] = temporal

print("Inventario ordenado por precio ascendente:")
for producto in inventario:
    print(f" - {producto['nombre']}: ${producto['precio']} | Stock: {producto['stock']}")
```

```python
import math


ancho = 5.0
alto = 3.0
rectangulo = (ancho, alto)


area = rectangulo[0] * rectangulo[1]
perimetro = 2 * (rectangulo[0] + rectangulo[1])

print(f"Dimensiones del rectángulo (ancho, alto): {rectangulo}")
print(f"Área: {area:.2f}")
print(f"Perímetro: {perimetro:.2f}")


try:
    rectangulo[0] = 10.0
except TypeError as error:
    print(f"\nError capturado al intentar cambiar 'ancho': {error}")


posiciones = [(0, 0), (1, 2), (3, 4)]
distancia_total = 0.0

for i in range(len(posiciones) - 1):
    x1 = posiciones[i][0]
    y1 = posiciones[i][1]

    x2 = posiciones[i + 1][0]
    y2 = posiciones[i + 1][1]

    distancia_tramo = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)
    distancia_total += distancia_tramo

print(f"\nDistancia total recorrida: {distancia_total:.4f}")
```

### Explicación de error

* **Error obtenido:**
  `TypeError: 'tuple' object does not support item assignment`

* **Explicación :**
  Las tuplas en Python son estructuras de datos **inmutables**. Esto significa que, una vez creadas en la memoria RAM, su tamaño y sus elementos no se pueden modificar, reemplazar ni eliminar.

* **Solución al problema:**
  Para alterar las dimensiones, es necesario crear una **nueva tupla** en memoria con los valores actualizados:
  rectangulo = (nuevo_ancho, alto)

```python

biblioteca = {
    "libros": [
        {"titulo": "1984", "autor": "George Orwell", "prestado": False},
        {"titulo": "El Principito", "autor": "Antoine de Saint-Exupéry", "prestado": True},
    ],
    "usuarios": ["Ana", "Luis", "María"]
}


libros_prestados = 0
libros_disponibles = 0

for libro in biblioteca["libros"]:
    if libro["prestado"]:
        libros_prestados += 1
    else:
        libros_disponibles += 1

print(f"Libros prestados: {libros_prestados}")
print(f"Libros disponibles: {libros_disponibles}")

nuevo_libro = {
    "titulo": "Cien años de soledad",
    "autor": "Gabriel García Márquez",
    "prestado": False
}
biblioteca["libros"].append(nuevo_libro)

for libro in biblioteca["libros"]:
    if libro["titulo"] == "1984":
        libro["prestado"] = True
        break

print("\nEstado actualizado de la biblioteca:")
print(biblioteca)
```
